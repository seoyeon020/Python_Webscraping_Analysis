#### 6. 함수

```python
# 문제 6-1: 두 수를 받아 사칙연산을 수행하는 함수 작성

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b
    
x, y = 10, 5
print(f'{x} + {y} = {add(x, y)}')
print(f'{x} - {y} = {subtract(x, y)}')
print(f'{x} * {y} = {multiply(x, y)}')
print(f'{x} / {y} = {divide(x, y)}')
```

```python
# 문제 6-2: 팩토리얼을 계산하는 재귀함수와 반복문을 사용한 함수

def recursive(n):
    if n <= 1: return 1
    return n * recursive(n-1)

def iterative(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

print(f'5! (재귀) = {recursive(5)}')
print(f'5! (반복) = {iterative(5)}')
print(f'7! (재귀) = {recursive(7)}')
print(f'7! (반복) = {iterative(7)}')
```

```python
# 문제 6-3: 기본값 매개변수를 사용하여 인사말 생성하는 함수

def greeting(name, greet="안녕하세요", say=""):
    if say == "hi":
        print(f'{greet}, {name}님! 좋은 하루 되세요!')
    elif greet == "Hello":
        print(f'{greet}, {name}!')
    else:
        print(f'{greet}, {name}님!')
        

greeting("김철수")
greeting("John", greet="Hello")
greeting("이영희", say="hi")
```

```python
# 문제 6-4: 리스트의 통계 정보(평균, 최댓값, 최솟값, 표준편차)를 반환하는 함수
import numpy as np

def list_info(numbers):
    print(f'숫자들: {numbers}')
    print(f'평균: {sum(numbers) / len(numbers)}')
    print(f'최댓값: {max(numbers)}')
    print(f'최솟값: {min(numbers)}')
    print(f'표준편차: {np.std(numbers):.2f}')

num = [10, 20, 30, 40, 50]
list_info(num)
```

```python
# 문제 6-5: 람다 함수를 사용하여 리스트를 다양한 방식으로 정렬

def lambda_sort(lists):
    print(f'학생 정보: {lists}')
    print(f'이름순 정렬: {sorted(lists, key=lambda x: x[0])}')
    print(f'점수순 정렬: {sorted(lists, key=lambda x: x[1])}')
    print(f'점수 내림차순: {sorted(lists, key=lambda x: x[1], reverse=True)}')


student = [('김철수', 85), ('이영희', 92), ('박민수', 78), ('최수진', 95)]
lambda_sort(student)
```
