#### 4. 제어문 처리하기

```python
# 문제 4-1: 점수를 입력받아 학점 출력

score = int(input("점수를 입력하세요: "))
print(f'점수를 입력하세요: {score}')

if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'F'

print(f'점수 {score}점의 학점은 {grade}입니다.')
```

```python
# 문제 4-2: 1부터 100까지 숫자 중 3의 배수의 합 구하기

num = []

for i in range(1, 101):
    if i % 3 == 0:
        num.append(i)

print(f'1부터 100까지 3의 배수: {num}')
print(f'3의 배수의 합: {sum(num)}')
print(f'3의 배수의 개수: {len(num)}개')
```

```python
# 문제 4-3: 구구단 출력

x = int(input("몇 단을 출력할까요? "))
print(f'몇 단을 출력할까요? {x}')
print(f'{x}단 구구단:')

for y in range(1, 10):
    print(f'{x} x {y} = {x*y}')
```

```python
# 문제 4-4: 숫자를 계속 입력밥아 0을 입력할때까지 합계 구하기

a = 1
a_sum = 0
while a != 0:
    a = int(input("숫자를 입력하세요 (0을 입력하면 종료): "))
    print(f'숫자를 입력하세요 (0을 입력하면 종료): {a}')
    a_sum += a

print(f'입력된 숫자들의 합: {a_sum}')
```

```python
# 문제 4-5: 입력한 숫자가 소수인지 판별

prime = int(input("숫자를 입력하세요: "))
print(f'숫자를 입력하세요: {prime}')

if prime > 1:
    for i in range(2, prime):
        if prime % i == 0:
            print(f'{prime}은 소수가 아닙니다.')
            break
    else:
        print(f'{prime}은 소수입니다.')
else:
    print("{prime}은 소수가 아닙니다.")
```
